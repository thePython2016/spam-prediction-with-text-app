```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import accuracy_score,precision_score,classification_report,f1_score,confusion_matrix
from sklearn.model_selection import train_test_split
import xgboost
from xgboost import XGBClassifier
```

```python
textSpam=pd.read_csv("spam.csv",encoding="ISO-8859-1")
textSpam
```

```python
textSpam=textSpam.drop(columns=['Unnamed: 2','Unnamed: 3','Unnamed: 4'])
```

```python
textSpam
```

```python
# reorder
textSpam=textSpam[['v2','v1']]
```

```python
textSpam.columns=["Text","is_fraud"]
```

```python
textSpam
```

```python
# cheeck null valuees
textSpam.isnull().sum()
```

```python
# Label encoder
labelEncoder=LabelEncoder()
```

```python
# split data
x=textSpam['Text']
y=textSpam['is_fraud']
xTrain,xTest,yTrain,yTest=train_test_split(x,y,test_size=0.25,random_state=34)
```

```python
# Encode Ytrain
yTrain=labelEncoder.fit_transform(yTrain)
```

```python
# encode ytest
yTest=labelEncoder.transform(yTest)
```

```python
# VECTORIZING TEXT
vector= TfidfVectorizer(
    max_features=3000,
    ngram_range=(1, 2),
    stop_words='english'
)

```

```python
 xTrain = vector.fit_transform(xTrain)
```

```python
xTest  = vector.transform(xTest)
```

```python
# model 
model = XGBClassifier(
    n_estimators=1000,
    learning_rate=0.1,
    reg_alpha=0.01,
    reg_lambda=0.01,
    random_state=42
)

model.fit(xTrain,yTrain)
```

```python
# prediction
featurePrediction=model.predict(xTest)
```

```python
labelEncoder.classes_
```

```python
labelEncoder.classes_=[]
```

```python
# classification report
report=classification_report(yTest,featurePrediction,target_names =labelEncoder.classes_,output_dict=True)
```

```python
# classification report Df
reportFrame=pd.DataFrame(report).transpose()
reportFrame
```
